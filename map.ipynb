```python
import pandas as pd
from pyproj import Transformer
import geopandas as gpd
from shapely.geometry import Point

df = pd.read_csv("./datasets/shelter/heat_wave.csv", encoding="euc-kr")
```

```python
geometry = [Point(xy) for xy in zip(df["경도"], df["위도"])]
gdf = gpd.GeoDataFrame(df, geometry=geometry, crs="epsg:4326")
```

```python
from geopy.distance import geodesic

user_location = Point(126.978, 37.5665)

radius = 3
filtered_points = [
    index
    for index, row in gdf.iterrows()
    if geodesic(
        (user_location.y, user_location.x), (row.geometry.y, row.geometry.x)
    ).kilometers
    <= radius
]
```

```python
import pandas as pd
from pyproj import Transformer

transformer = Transformer.from_crs("epsg:5186", "epsg:4326")
df = pd.read_csv("./datasets/shelter/fine_dust.csv", encoding="euc-kr")
lat, lon = transformer.transform(df["Y좌표"].values, df["X좌표"].values)
```

```python
df["경도"] = lon
df["위도"] = lat
```

```python
df.to_csv("./datasets/shelter/fine_dust_prepared.csv", index=False)
```
