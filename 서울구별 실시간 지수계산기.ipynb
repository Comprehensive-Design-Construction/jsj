```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# 드라이버 설정
chrome_driver_path = r"C:\chromedriver\chromedriver.exe"
options = Options()
options.add_experimental_option("detach", True)
service = Service(executable_path=chrome_driver_path)
driver = webdriver.Chrome(service=service, options=options)

# 페이지 열기
url = "https://www.weather.go.kr/w/index.do#dong/1150062000/37.556107/126.812164/서울특별시%20강서구%20공항동/SEL/공항동"
driver.get(url)

try:
    # 최대 15초까지 대기하며 습도 값을 찾기
    humidity_element = WebDriverWait(driver, 15).until(
        EC.presence_of_element_located(
            (By.XPATH, "//div[contains(@class,'cmp-main-aws')]//table//tr[1]/td[3]")
        )
    )
    humidity = humidity_element.text.strip()
    print("🌫 현재 습도:", humidity)

except Exception as e:
    print("❌ 습도 값을 가져오는 데 실패했습니다:", e)
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# 드라이버 설정
chrome_driver_path = r"C:\chromedriver\chromedriver.exe"
options = Options()
options.add_experimental_option("detach", True)
options.add_argument('headless')
options.add_argument('window-size=1920x1080')
options.add_argument("disable-gpu")
service = Service(executable_path=chrome_driver_path)
driver = webdriver.Chrome(service=service, options=options)

# 페이지 열기
url = "https://www.weather.go.kr/w/index.do#dong/1150062000/37.556107/126.812164/서울특별시%20강서구%20공항동/SEL/공항동"
driver.get(url)

try:
    wait = WebDriverWait(driver, 15)

    # 습도
    humidity_element = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//div[contains(@class,'cmp-main-aws')]//table//tr[1]/td[3]")
        )
    )
    humidity = humidity_element.text.strip()
    print("🌫 현재 습도:", humidity)

    # 최저 기온
    low_temp_element = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//div[@class='daily-minmax']/div[1]/span")
        )
    )
    low_temp = low_temp_element.text.strip()
    print("🌡 오늘 최저 기온:", low_temp)

    # 최고 기온
    high_temp_element = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//div[@class='daily-minmax']/div[2]/span")
        )
    )
    high_temp = high_temp_element.text.strip()
    print("🔥 오늘 최고 기온:", high_temp)

except Exception as e:
    print("❌ 데이터를 가져오는 데 실패했습니다:", e)
```
